STEP 3

```python
df_clean = df.copy()

print("Original shape:", df.shape)
print("Working shape:", df_clean.shape)
```

```python
df_clean = df_clean.drop_duplicates()

print("Shape after removing exact duplicates:", df_clean.shape)
```

```python
text_columns = ["flight_id", "airline", "source", "destination"]

for col in text_columns:
    df_clean[col] = df_clean[col].astype("string").str.strip()

print(df_clean[text_columns].head())
```

```python
df_clean["airline"] = df_clean["airline"].fillna("UNKNOWN")

print(df_clean["airline"].value_counts(dropna=False))
print("\nMissing airline values:", df_clean["airline"].isna().sum())
```

```python
df_clean["duration_clean"] = df_clean["duration"].apply(
    lambda x: pd.Timedelta(
        hours=x.hour,
        minutes=x.minute,
        seconds=x.second
    ) if isinstance(x, datetime.time)
    else pd.Timedelta(
        days=x.day - 1,
        hours=x.hour,
        minutes=x.minute,
        seconds=x.second
    )
)

print(df_clean["duration_clean"].head())
print("\nDuration data type:", df_clean["duration_clean"].dtype)
```

```python
invalid_time = df_clean["arrival_time"] < df_clean["departure_time"]

df_clean.loc[invalid_time, "arrival_time"] = (
    df_clean.loc[invalid_time, "arrival_time"] + pd.Timedelta(days=1)
)

print("Invalid timestamp orders remaining:",
      (df_clean["arrival_time"] < df_clean["departure_time"]).sum())
```

```python
df_clean["calculated_duration"] = (
    df_clean["arrival_time"] - df_clean["departure_time"]
)

print(df_clean[[
    "flight_id",
    "departure_time",
    "arrival_time",
    "duration_clean",
    "calculated_duration"
]].head())
```

```python
duration_match = (
    df_clean["duration_clean"] == df_clean["calculated_duration"]
)

print("Matching durations:", duration_match.sum())
print("Non-matching durations:", (~duration_match).sum())
```

```python
print("Final shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isna().sum())

print("\nExact duplicates:", df_clean.duplicated().sum())

print("\nInvalid timestamp orders:",
      (df_clean["arrival_time"] < df_clean["departure_time"]).sum())
```

```python
mismatches = df_clean[
    df_clean["duration_clean"] != df_clean["calculated_duration"]
]

print("Number of mismatches:", len(mismatches))

mismatches[[
    "flight_id",
    "airline",
    "source",
    "destination",
    "departure_time",
    "arrival_time",
    "duration_clean",
    "calculated_duration"
]].head(30)
```

```python
df_clean["flight_duration"] = (
    df_clean["calculated_duration"].dt.total_seconds() / 60
).round().astype(int)

print(df_clean[[
    "flight_id",
    "duration_clean",
    "calculated_duration",
    "flight_duration"
]].head())
```

```python
print("Minimum duration:", df_clean["flight_duration"].min(), "minutes")
print("Maximum duration:", df_clean["flight_duration"].max(), "minutes")

print("Invalid durations:",
      (df_clean["flight_duration"] <= 0).sum())
```

```python
print("Unique sources:", df_clean["source"].unique())
print("Unique destinations:", df_clean["destination"].unique())

same_route = (df_clean["source"] == df_clean["destination"]).sum()
print("Same source/destination flights:", same_route)
```

```python
duplicate_ids = df_clean[
    df_clean["flight_id"].duplicated(keep=False)
].sort_values("flight_id")

print("Rows belonging to duplicate flight IDs:", len(duplicate_ids))

duplicate_ids.head(30)
```

```python
df_clean["route"] = (
    df_clean["source"] + " → " + df_clean["destination"]
)

print(df_clean[["source", "destination", "route"]].head())
```

```python
df_clean["airline_standardized"] = (
    df_clean["airline"]
    .str.upper()
    .str.strip()
)

print(df_clean["airline_standardized"].value_counts())
```

```python
df_clean["departure_date"] = df_clean["departure_time"].dt.date

print(df_clean[["departure_time", "departure_date"]].head())
```

```python
df_clean["is_overnight"] = (
    df_clean["departure_time"].dt.date
    != df_clean["arrival_time"].dt.date
)

print(df_clean["is_overnight"].value_counts())
```

```python
df_clean["flight_duration_hours"] = (
    df_clean["flight_duration"] / 60
).round(2)

print(
    df_clean[
        ["flight_id", "flight_duration", "flight_duration_hours"]
    ].head()
)
```

```python
print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))

print("\nMissing values:")
print(df_clean.isna().sum().sum())

print("\nExact duplicates:", df_clean.duplicated().sum())

print(
    "\nInvalid timestamps:",
    (df_clean["arrival_time"] < df_clean["departure_time"]).sum()
)

print(
    "Invalid durations:",
    (df_clean["flight_duration"] <= 0).sum()
)

print(
    "Overnight flights:",
    df_clean["is_overnight"].sum()
)

print("\nFinal columns:")
print(df_clean.columns.tolist())
```

```python
avg_duration = df_clean["flight_duration_hours"].mean()

print(f"Average Flight Duration: {avg_duration:.2f} hours")
```

```python
route_traffic = (
    df_clean.groupby("route")
    .size()
    .reset_index(name="flight_count")
    .sort_values("flight_count", ascending=False)
)

route_traffic
```

```python
airline_distribution = (
    df_clean.groupby("airline_standardized")
    .size()
    .reset_index(name="flight_count")
    .sort_values("flight_count", ascending=False)
)

airline_distribution
```

```python
airline_duration = (
    df_clean.groupby("airline_standardized")["flight_duration_hours"]
    .mean()
    .reset_index(name="average_duration_hours")
    .sort_values("average_duration_hours", ascending=False)
)

airline_duration
```

```python
anomaly_summary = {
    "Missing/Unknown Airlines": (df_clean["airline_standardized"] == "UNKNOWN").sum(),
    "Overnight Flights": df_clean["is_overnight"].sum(),
    "Duplicate Flight IDs": df_clean["flight_id"].duplicated().sum()
}

anomaly_summary
```
